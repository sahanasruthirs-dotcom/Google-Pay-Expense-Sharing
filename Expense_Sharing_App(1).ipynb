```python
import numpy as np
from prettytable import PrettyTable
```

```python
friends = ["Alice", "Bob", "Carol", "David"]

expense_matrix = np.zeros(
    (len(friends), len(friends))
)

transaction_history = []

print("Friends Added Successfully")
```

```python
def display_matrix():

    table = PrettyTable()

    table.field_names = [""] + friends

    for i in range(len(friends)):
        row = [friends[i]]

        for j in range(len(friends)):
            row.append(
                round(expense_matrix[i][j],2)
            )

        table.add_row(row)

    print(table)
```

```python
def add_expense(
    payer,
    beneficiaries,
    amount
):

    payer_idx = friends.index(payer)

    share_per_person = (
        amount / len(beneficiaries)
    )

    for beneficiary in beneficiaries:

        beneficiary_idx = friends.index(
            beneficiary
        )

        expense_matrix[
            payer_idx
        ][
            beneficiary_idx
        ] += share_per_person

    transaction_history.append({
        "Type":"Equal",
        "Payer":payer,
        "Beneficiaries":beneficiaries,
        "Amount":amount
    })

    print("Expense Added")
```

```python
def add_custom_expense(
    payer,
    custom_shares
):

    payer_idx = friends.index(payer)

    total = sum(
        custom_shares.values()
    )

    for person, share in custom_shares.items():

        person_idx = friends.index(
            person
        )

        expense_matrix[
            payer_idx
        ][
            person_idx
        ] += share

    transaction_history.append({
        "Type":"Custom",
        "Payer":payer,
        "Shares":custom_shares,
        "Amount":total
    })

    print("Custom Expense Added")
```

```python
def show_history():

    table = PrettyTable()

    table.field_names = [
        "Type",
        "Payer",
        "Details"
    ]

    for transaction in transaction_history:

        table.add_row([
            transaction["Type"],
            transaction["Payer"],
            str(transaction)
        ])

    print(table)
```

```python
def calculate_settlements():

    total_paid = np.sum(
        expense_matrix,
        axis=1
    )

    total_owed = np.sum(
        expense_matrix,
        axis=0
    )

    net_balance = (
        total_paid - total_owed
    )

    table = PrettyTable()

    table.field_names = [
        "Friend",
        "Balance"
    ]

    for i in range(
        len(friends)
    ):

        table.add_row([
            friends[i],
            round(
                net_balance[i],
                2
            )
        ])

    print("\nSettlement Summary")
    print(table)

    return net_balance
```

```python
def show_balances():

    balances = calculate_settlements()

    print("\nBalance Report\n")

    for i in range(len(friends)):

        if balances[i] > 0:

            print(
                f"{friends[i]} should receive ₹{balances[i]:.2f}"
            )

        elif balances[i] < 0:

            print(
                f"{friends[i]} owes ₹{abs(balances[i]):.2f}"
            )

        else:

            print(
                f"{friends[i]} is settled"
            )
```
